```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def binary_step(x):
    return np.where(x >= 0.5, 1, 0)

# AND Gate Inputs and Targets
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_true = np.array([[0], [0], [0], [1]])

# Weights and Bias for AND gate
# High weights and a negative bias to ensure only (1,1) triggers the output
weights = np.array([[20], [20]])
bias = -30

# Single Layer Forward Propagation
linear_output = np.dot(X, weights) + bias
probabilities = sigmoid(linear_output)
final_output = binary_step(probabilities)

print("Inputs:")
print(X)
print("\nPredicted Probabilities:")
print(probabilities)
print("\nFinal Binary Output (AND Gate):")
print(final_output)

# Calculate Loss
loss = np.mean(np.power(y_true - probabilities, 2))
print(f"\nMean Squared Error Loss: {loss:.6f}")
```
