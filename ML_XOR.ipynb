```python
import math
import random

# Sigmoid Activation Function
def sigmoid(x):
    return 1 / (1 + math.exp(-x))

# Generating Random XOR Inputs
print("Random XOR Input:")

x1 = random.choice([0, 1])
x2 = random.choice([0, 1])

print("x1 =", x1)
print("x2 =", x2)

# Hidden Layer Weights
print("\nRandom Weights for Hidden Layer")

w11 = random.uniform(-5, 5)
w21 = random.uniform(-5, 5)

w12 = random.uniform(-5, 5)
w22 = random.uniform(-5, 5)

# Hidden Layer Biases
b1 = random.uniform(-5, 5)
b2 = random.uniform(-5, 5)

# Output Layer Weights
print("\nRandom Weights for Output Layer")

v1 = random.uniform(-5, 5)
v2 = random.uniform(-5, 5)

# Output Bias
b3 = random.uniform(-5, 5)

# Display Random Weights and Biases
print("\nGenerated Weights and Biases:")
print("w11 =", round(w11, 4))
print("w21 =", round(w21, 4))
print("w12 =", round(w12, 4))
print("w22 =", round(w22, 4))
print("b1 =", round(b1, 4))
print("b2 =", round(b2, 4))
print("v1 =", round(v1, 4))
print("v2 =", round(v2, 4))
print("b3 =", round(b3, 4))

# ---------- Forward Propagation ----------

# Hidden Layer
Zh1 = x1*w11 + x2*w21 + b1
Zh2 = x1*w12 + x2*w22 + b2

h1 = sigmoid(Zh1)
h2 = sigmoid(Zh2)

# Output Layer
Z0 = h1*v1 + h2*v2 + b3
output = sigmoid(Z0)

# Final Prediction
prediction = 1 if output >= 0.5 else 0

# Display Results
print("\n----------- RESULT -----------")
print("Hidden Neuron h1 =", round(h1, 4))
print("Hidden Neuron h2 =", round(h2, 4))
print("Output =", round(output, 4))
print("Predicted XOR =", prediction)
```
